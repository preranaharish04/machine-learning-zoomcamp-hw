```python
median_before = df['horsepower'].median()
print('Median before:', median_before)

mode_hp = df['horsepower'].mode()[0]
print('Most frequent value:', mode_hp)

df['horsepower'] = df['horsepower'].fillna(mode_hp)

median_after = df['horsepower'].median()
print('Median after:', median_after)

if median_after > median_before: print('Yes, it increased')
elif median_after < median_before: print('Yes, it decreased')
else: print('No')
```

```python
asia = df[df['origin'] == 'Asia']
X = asia[['vehicle_weight', 'model_year']].head(7).values
print(X)

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y
print('w =', w)
print('Sum =', w.sum())
```
